## Subset Sum Problem
The Subset Sum Problem (SSP) is a NP-complete problem in computer science. It can be formulated as a decision problem or as an optimization problem. The decision problem formulation is the one this algorithm is bulit for, therefore is the formulation we will be using in this notebook.
$$
\text{Given a set}\, A \,\text{and a target}\, t \, \text{, does there exists a subset}\, S\subset A \,\text{such that}\, \sum_{s \in S}s=t\, \text{?}
$$

Classical solutions for the SSP include brute-force, dynamic programming approaches. Quantum solutions include quantum random-walk algorithms, QUBO forumulation & Ising models simulations and Grover's search algorithm, which is the one we present here.

Let's start by defining our problem. Suppose that we have a set A of integers: $A\subset \mathbb{Z}$ and a target integer: $t\in \mathbb{Z}$. The goal is to find if a subset $S$ of $A$ ($S \subset A$) verifies the problem statement, that is if all of the elements $s$ of subset $S$ sum to target $t$. 
- If such a subset exists, the instance is solved and the subset is shown. 
- If it doesn't exists, the instance doesn't have a solution. 
- If there are more than one subset that solve an instance, then is has multiple solutions.

It is easy to see that the problem is reduced to a unstructured search instance where we must generate all of the possible subsets $S$ of set $A$ with the value of the additions of their elements. Our approach for this problem uses two main algorithms for generating the solution space and executing a quantum search: 
- Draper's adder
- Grover's search algorithm

### Draper's Adder

Draper's Adder is a quantum addition algorithm created by Thomas G. Draper: [Addition on a Quantum Computer](https://arxiv.org/abs/quant-ph/0008033). The main difference betwqeen Draper's adder and other designs is that it uses the Quantum Fourier Transform to convert the computational basis states of the binary expression of the first number to add into the Fourier basis (phase rotations) to perform the addition. This represents a mayor difference between Draper's addition and other addition models, because it eliminates the need for additional qubits. Although, it comes at the cost of implementing the QFT operation which is expensive to do so.

### Grover's search algorithm

Grover's search algorithm deals with the problem of unstructed search through a dataset. For a search query through a list of size $N$, Grover's algorithm only requires $\sqrt{N}$ steps to find a specific element, therefor it has a temporal complexity of $\mathcal{O}(n)$.
