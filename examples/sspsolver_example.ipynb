# SSP Solver: Usage Example

This notebook provides a detailed example for how to solve a Subset Sum Problem (SSP) instance usign the Draper+Grover algorithm. Let's start by reminding what is the Subset Sum Problem and how is it defined.

This notebook demonstrates how to use the `ssp-solver` library to:

1. Define an SSP instance  
2. Build and visualize the Grover-based quantum circuit  
3. Run it on a simulator  
4. Process the measurements for the most likely solutions  
5. Inspect basic circuit stats  

## Subset Sum Problem
The Subset Sum Problem (SSP) is a NP-hard problem in computer science. It can be formulated as a decision problem or as an optimization problem. The decision problem forumulation is the one this algorithm is bulit for, therefor is the formulation we will be using in this notebook.
